```python
#read webcam
import cv2
webcam = cv2.VideoCapture(0)  #2 is for all web cam
```

```python
from google.colab.patches import cv2_imshow
#visualize webcam
while True:
  ret, frame = webcam.read()
  if not ret:
    print("Failed to grab frame, exiting...")
    break
  cv2_imshow(frame)
  if cv2.waitKey(500) & 0xFF == ord('q'):
    break
webcam.release()
cv2.destroyAllWindows() # Fixed typo here
```

```python
#colour spaces
import cv2
import os
from google.colab.patches import cv2_imshow
img = cv2.imread(os.path.join('.','data','/content/blur.jpg'))

# Check if image was loaded successfully before proceeding
if img is None:
    print("Error: Could not load image 'mountain.jpg'. Please ensure it's in the ./data/ directory.")
else:
    img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

    # Display images using cv2_imshow without window names
    cv2_imshow(img)
    cv2_imshow(img_gray)
    cv2_imshow(img_rgb)
    cv2_imshow(img_hsv)

    # cv2.waitKey(0) is not typically used with cv2_imshow in Colab as it's for interactive windows
    # The outputs will be displayed in the cell output without needing to wait.
```

```python
#Blur
import os
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread(os.path.join('.','data','/content/blur.jpg'))
k_size  = 10
img_blur = cv2.blur(img, (k_size, k_size))

cv2_imshow(img)
cv2_imshow(img_blur)
# cv2.waitKey(0) is not typically used with cv2_imshow in Colab
```

```python
#GaussianBlur
#medium blur
import os
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread(os.path.join('.','data','/content/blur.jpg'))
k_size  = 11 # Changed k_size to an odd number for GaussianBlur
img_blur = cv2.blur(img, (10, 10)) # Retained original k_size for cv2.blur if needed differently
img_gaussian_blur = cv2.GaussianBlur(img, (k_size, k_size), 3)
img_median_blur = cv2.medianBlur(img, k_size)
cv2_imshow(img)
cv2_imshow(img_blur)
cv2_imshow(img_gaussian_blur)
cv2_imshow(img_median_blur)
```

```python
#Thresholding for image
import os
import cv2
from google.colab.patches import cv2_imshow # Corrected import
img = cv2.imread(os.path.join('.','data','/content/blur.jpg'))

if img is None:
    print("Error: Could not load image. Please ensure '/content/blur.jpg' is correct.")
else:
    img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    # Ensure the thresholding operation is applied correctly, usually it returns two values
    # The second value is the thresholded image. It is good practice not to modify img_gray in place if it's used elsewhere.
    ret, img_thresholded = cv2.threshold(img_gray, 100, 150, cv2.THRESH_BINARY)

    cv2_imshow(img) # Use cv2_imshow
    cv2_imshow(img_gray) # Display grayscale image
    cv2_imshow(img_thresholded) # Display thresholded image
    # cv2.waitKey(0) is not typically used with cv2_imshow in Colab
```
